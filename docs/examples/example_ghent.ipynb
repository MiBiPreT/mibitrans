# Field site example: tar and soot factory Ghent

This notebooks outlines analysis of a contaminated fieldsite in *Ghent*, Belgium. The focus is on the flexibility of mibitrans, regarding both the limited information available for the site and the testing of various parameter scenarios for a location.

**Author:** Noortje van Strien

## Site description

A tar and soot factory was formerly located in the Wondelgem district of Ghent.$^1$ Industrial processes and incidents resulted in soil contamination consisting of a mixture of BTEX, C6-C10 (light mineral oil fraction), and various PAH components.$^2$ The factory was located approximately 50 meters from the Lieve Canal, meaning that contamination on the site led to pollution of the canal's surface water.

In 2019, heavily contaminated sediment was removed from the Lieve, but this resulted in increased inflow into the canal. Consequently, a reactive mat made of absorbent materials was installed in the canal in 2020 to improve surface water quality.$^1$

Flow and contaminant transport occurs within an unconfined aquifer, with groundwater levels ranging from 1.5 and 2.2 meters below ground level. The upper five meters consits of fine, silty sand with groundwater flowing towards the Lieve Canal; towards the southwest. Underlying this are layers of medium-coarse sand interspersed with loam layers, where the groundwater flows away from the Lieve Canal; towards the northeast.

## Model setup
The aim is to set up a transport model for the Ghent site based on site characteristics and to compare the results with on-site measurements. Due to limited information about the location and the associated uncertainties, multiple scenarios will be implemented. This will be carried out for the BTEX component of the source.

### Baseline scenario
The baseline scenario is a starting point that uses a realistic set of parameter values within the current uncertainties. In the following scenarios, these parameters are varied to see whether different parameter values can still produce results that match the field measurements. 

#### Hydrogeological parameters
The horizontal groundwater velocity was determined using iFLUX, reseulting in a flow velocity of $10.5$ $cm/d$ with a range of $9-12$ $cm/d$.$^1$ Additionally, a porosity of 0.3 was established. No field data regarding dispersivity values for the site where available. Therefore, these were estimated based on $[4,5]$, leading to the estimated values $\alpha_L = 2.3$ $m$, $\alpha_T= 0.02$ $m$ and $\alpha_V=0.004$ $m$.

#### Attenuation parameters
A retardation factor of $3.04$ is used $^6$. This is calculated by applying the proportions of the BTEX components in the source to their respective retardation factors. A decay rate of $0.025$ $d^{-1}$ was estimated $^6$, but in this example a decay rate of $0$ $d^{-1}$ will be implemented.

#### Source conditions
The source thickness is estimated to be roughly $15$ meters. The total source width is $66$ meters, with a $28$ meter zone of $2$ $g/m^3$, a $10$ meter zone of $54.7$ $g/m^3$ and a $28$ meter zone of $2$ $g/m^3$. $^2$ Note that these distances and concentrations have been estimated based on available field measurements, given the lack of information regarding the source.

#### Model parameters
The model dimensions are based on the position of the source and the position and timing of the field measurements, and are as follows:
- Length = $50$ meters
- Width = $120$ meters
- Duration = $790$ days = $2.16$ years

### Dispersivity scenarios
Two scenarios are evaluated in which the transverse horizontal dispersivity is modified and compared with the baseline scenario, while all other parameters remain identical to those of the baseline scenario. All values fall within the range of possible values for transverse horizontal dispersivity. However, the transverse horizontal dispersivity used in the reference scenario has high reliability $^{4,5}$, whereas the values in scenarios d1 and d2 have moderate reliability.
- Scenario d1: $\alpha_T= 0.1$ meter
- Scenario d2: $\alpha_T= 0.001$ meter

### Source dimensions scenarios
Source dimensions represent one of the major uncertainties for this site. Therefore, scenarios with varying source widths for the inner source zone are being modeled to determine the potential range of this zone's width. All other parameters remain identical to those of the reference scenario. The widths of the inner source zone will be:
- Scenario s0: $6$ meter
- Scenario s1: $30$ meter
- Scenario s2: $50$ meter

### Combined variation scenario
This scenario (Scenario cv) involves source concentrations, source widths and transverse horizontal dispersivity that differ from those in the base scenario, while all other parameters remain identical. The source is divided into three zones, rather than the two used in the previous scenarios. The concentrations used are $[62, 20, 2]$ $g/m^3$ and the source dimensions are $[4, 10, 28]$ meters. The transverse horizontal dispersivity is set at $0.2$ meters. This is on the higher end, but still realistic. $^{4,5}$

```python
import matplotlib.pyplot as plt
import numpy as np
import mibitrans as mbt
```

## Field Data

```python
field_data_Ghent_x = np.array([-30, 0, 40 , 50])
field_data_Ghent_c = np.array([4.62*0.001, 8750*0.001, 45.21*0.001, 39.46*0.001])
```

```python
plt.figure(figsize = [6,3])
plt.plot(field_data_Ghent_x, field_data_Ghent_c, color="black", marker = 'o', ms=10, ls = ':')
plt.title('Ghent field observation on contaminant concentration')
plt.show()
```

## Baseline scenario
### Setup Data Input

```python
hydro = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.02, #[m]
    alpha_z = 0.004 #[m]
)
```

```python
att = mbt.AttenuationParameters(
    decay_rate = 0, #[1/day]
    retardation = 3.04 #[-]
)
```

```python
source = mbt.SourceParameters(
    source_zone_boundary = np.array([5,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)
```

```python
model = mbt.ModelParameters(
    model_length = 50, #[m]
    model_width = 120, #[m]
    model_time = 790, #[days]
    dx = 1, #[m]
    dy = 1, #[m]
    dt = 79, #[days]
)
```

### Modelling

```python
mbt_ghent = mbt.Mibitrans(hydro, att, source, model)
results_ghent = mbt_ghent.run()
```

### Plotting
#### Concentration along plume center over time

```python
cmap = plt.get_cmap("cubehelix")
colors = cmap(np.linspace(0.4, 0.7, 5))
```

```python
for i,plot_time in enumerate([7.9, 7, 5, 3, 1]):
    results_ghent.centerline(time=plot_time*100,
                               color=colors[i],
                               lw = 2,
                               ls = "-",
                               label=f"t={plot_time*100}days")
plt.legend()
plt.show()
```

**Interpretation**: Over time, the contamination migrates further into the model domain, causing the concentration to increase. No steady state is reached, as the decay rate is set to $0$ $d^{-1}$.

#### Comparison to field measurements

```python
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: The field measurements align well with the mibitrans results. Although there are some minor differences, the results follow the same trend as the field measurements. However, due to the lack of field data between 0 and 30 meters on either side, and because of parameter uncertainties, there are multiple possible parameter combinations that represent this field data well.

## Dispersivity scenarios

```python
# Setup data input
hydro_d1 = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.1, #[m]
    alpha_z = 0.004 #[m]
)

hydro_d2 = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.001, #[m]
    alpha_z = 0.004 #[m]
)
```

```python
# Modelling
mbt_ghent_d1 = mbt.Mibitrans(hydro_d1, att, source, model)
results_ghent_d1 = mbt_ghent_d1.run()

mbt_ghent_d2 = mbt.Mibitrans(hydro_d2, att, source, model)
results_ghent_d2 = mbt_ghent_d2.run()
```

```python
# Plotting
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
results_ghent_d1.transverse(x_position= 40, lw = 2, ls = "-", color = colors[2],  label = "d1 scenario")
results_ghent_d2.transverse(x_position= 40, lw = 2, ls = "-", color = colors [4], label = "d2 scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: Although the shape of the concentration distribution differs across the three scenarios, they align with the field measurements in a similar manner. Since the values for transverse horizontal dispersivity are representative in all three scenarios, the value can lie anywhere within this plausible range of 0.001 to 0.1 meters.

**Note**: The entire model does not need to be recreated. Only the data class in which the parameter changes needs to be redefined. During modeling, the original data classes can be provided as input alongside the modified data class.

## Source dimensions scenarios

```python
# Source data input
source_s0 = mbt.SourceParameters(
    source_zone_boundary = np.array([3,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)

source_s1 = mbt.SourceParameters(
    source_zone_boundary = np.array([15,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)

source_s2 = mbt.SourceParameters(
    source_zone_boundary = np.array([25,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)
```

```python
# Modelling
mbt_ghent_s0 = mbt.Mibitrans(hydro, att, source_s0, model)
results_ghent_s0 = mbt_ghent_s0.run()

mbt_ghent_s1 = mbt.Mibitrans(hydro, att, source_s1, model)
results_ghent_s1 = mbt_ghent_s1.run()

mbt_ghent_s2 = mbt.Mibitrans(hydro, att, source_s2, model)
results_ghent_s2 = mbt_ghent_s2.run()
```

```python
# Plotting
results_ghent_s0.transverse(x_position= 40, lw = 2, ls = "-", color = colors[1], label = "s0 scenario")
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
results_ghent_s1.transverse(x_position= 40, lw = 2, ls = "-", color = colors[3], label = "s1 scenario")
results_ghent_s2.transverse(x_position= 40, lw = 2, ls = "-", color = colors[4], label = "s2 scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: In these scenarios, the values for the width of the inner source zone range from 6 to 50 meters. This is a very wide range, resulting in vastly different sources for the different scenarios. However, all these scenarios align well with the field measurements and must therefore be considered, as they may well be correct.

## Combined variation scenario

```python
# Source data input
source_cv = mbt.SourceParameters(
    source_zone_boundary = np.array([4,10,28]), #[m]
    source_zone_concentration = np.array([62,20,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)

hydro_cv = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.2, #[m]
    alpha_z = 0.004 #[m]
)
```

```python
# Modelling
mbt_ghent_cv = mbt.Mibitrans(hydro_cv, att, source_cv, model)
results_ghent_cv = mbt_ghent_cv.run()
```

```python
#Plotting
results_ghent_cv.transverse(x_position= 40, lw = 2, ls = "-", color = colors[4], label = "cv scenario")
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: Of all the scenarios, the shape of the graph for this scenario deviates most strongly from that of the baseline scenario. Nevertheless, this graph still aligns with the field measurements. Various combinations of the uncertain parameters can also yield representative scenarios.

```python
# Plotting
results_ghent_cv.centerline(lw = 2, ls = "-", color = colors[4], label = "cv scenario")
results_ghent.centerline(lw = 2, ls = "-", color = colors[0], label = "baseline scenario")
plt.ylim(0,65)
plt.legend()
plt.show()
```

**Interpretation**: 
In the previous scenarios, the graphs for the centerline remained the same, with differences occurring only in the graphs for the transverse direction. In this scenario, however, the centerline graph also deviates. This is because the source zones have a different concentration than in all previous scenarios. The difference in concentration between the scenarios occurs only within the first 30 meters. Beyond that point, the lines coincide, explaining why the requirement for the middle measurement point is still met.

**Conclusion**: All scenarios demonstrated that varying parameter values for the greatest uncertainties, in this case regarding dispersivity and the source, or a combination thereof, yield results consistent with field measurements. When facing such uncertainties, mibitrans can be useful for rapidly determining the potential range of uncertain parameters by setting up various scenarios, as demonstrated in this notebook. Moreover, it enables an accurate model setup even when only limited information is available.
