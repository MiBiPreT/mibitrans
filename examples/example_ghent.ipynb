# Field site example: tar and soot factory Ghent

This notebooks outlines analysis of a contaminated fieldsite in *Ghent*, Belgium. The focus is on the flexibility of mibitrans, regarding both the limited information available for the site and the testing of various parameter scenarios for a location.

**Author:** Noortje van Strien and Alraune Zech

## Site description

A tar and soot factory was formerly located in the Wondelgem district of Ghent$^1$. Industrial processes and incidents resulted in contamination of the upper soil with a mixture of BTEX, C6-C10 (light mineral oil fraction), and various PAH components$^2$. The factory was located approximately 50 meters from the Lieve Canal. This canal drains to the Ghent–Terneuzen Canal. Pure phase flow and groundwater flow towards the canal caused pollution of the canal's surface water.

In 2019, a remediation operation was executed where heavily contaminated sediment was removed from the Lieve. However, the change in sediment permeability resulted in an increase of inflow of contaminanted groundwater into the canal. Consequently, a reactive mat made of absorbent materials was installed in the canal in 2020 to improve surface water quality$^1$. 

Flow and contaminant transport occurs within an unconfined aquifer, with groundwater levels ranging from 1.5 and 2.2 meters below ground level. Groundwater flows towards the Lieve Canal; in south-west direction. The upper five meters of the aquifer consist of fine, silty sand. Underlying this are layers of medium-coarse sand interspersed with loam layers, where the groundwater flows away from the Lieve Canal; towards the north-east.

## Model setup
The notebook presents a transport model for the Ghent site based on site characteristics to compare with on-site measurements. 

The model reflects the transport of BTEX contaminants from an extended source location 50m upstream of the Lieve Canal  toward the Lieve Canal (receptor). Advective transport takes place in the main direction of groundwater flow that is perpendicular to the Lieve Canal. 

Availability of flow and transport parameters from field observations is limited.  We define a baseline scenario using a set of parameter values that are reconstructed from literature and observations. To account for the uncertainty associated with the limited information, we define multiple scenarios, testing a range of parameters that are higly uncertain, including (i) dispersivity scenarios, (ii) source dimensions scenarios and (iii) combined variation scenario.

### Source conditions
Source dimensions are only roughly known$^2$. As base scenario, we use an plane source at 50m distance to the Lieve Canal with a total width of $66$ and a thickness of $15$ meters. The central point of the contaminant source is located upstream of *Oip-5*, given the high values of contamination at *Oip-5* and historical specification of the contaminant source location. 

The total source width of $66$m contains different zones: (i) a central zone of $10$ meter width with a contaminant concentration of $54.7$ $g/m^3$, (ii) and two secondary zones (left and right to the central zone) of each $28$ m width with a contaminant concentration of $2$ $g/m^3$. Source concentrations have been estimated based on available field measurements.

### Model domain parameters

Following the model setup and the source specification, we use:
- a domain length of $50$ meters ($x$-direction in direction of flow from source to Canal)
- a domain width of $120$ meters ($y$-direction to account for extended source width)
- a model duration of $790$ days ($2.16$ years)
  
The extended model duration is chosen according to the timing of the field measurements after removing the sediments and the installation of the reactive mat.


### Hydrogeological parameters
The horizontal groundwater velocity was determined using iFLUX$^1$, resulting in a range of $9-12$ $cm/d$. We use a flow velocity of $v=10.5$ $cm/d$. Porosity was estimated to be $0.3$. 

No field data regarding dispersivity values for the site where available. We developed realistic estimates based on $[4,5]$ of: longitudinal dispersivity $\alpha_L = 2.3$ $m$, transverse horizontal dispersivity $\alpha_T= 0.02$ $m$ and transverse vertical dispersivity $\alpha_V=0.004$ $m$.

### Attenuation parameters
A retardation factor of $R = 3.04$ is used following$^6$. The value was calculated by applying the proportions of the BTEX components in the source to their respective retardation factors. The decay rate is set to $0$ in this example.


## Model scenarios

### Dispersivity scenarios

We run and evaluate two scenarios with alternative values of transverse horizontal dispersivity $\alpha_T$. The value of $\alpha_T= 0.02$ $m$ used in the baseline scenario is taken from $[4,5]$ citing measured field values of high reliability. However, other field values (of moderate reliability) are available. We test the values of

- Scenario d1: $\alpha_T= 0.1$ meter
- Scenario d2: $\alpha_T= 0.001$ meter

All other model parameters remain identical to those of the baseline scenario. 

### Source dimensions scenarios

Source dimensions represent one of the major uncertainties for this site. We test three scenarios with varying source widths for the center source zone (of high contaminant concentration) to study the impact and determine the potential range of this zone's width. The tested widths are:

- Scenario s0: $6$ meter
- Scenario s1: $30$ meter
- Scenario s2: $50$ meter
 
Again, all other parameters remain identical to those of the baseline scenario.

### Combined variation scenario

This scenario (scenario cv) combines alternative source concentrations, source widths and transverse horizontal dispersivity (compared to the base scenario. All other model parameters remain identical. The source is divided into three (rather than two) zones with dimensions of $[4, 10, 28]$ meters and concentrations of $[62, 20, 2]$ $g/m^3$. The transverse horizontal dispersivity is set to $0.2$ meters.

```python
import matplotlib.pyplot as plt
import numpy as np
import mibitrans as mbt
```

## Field Data

```python
field_data_Ghent_x = np.array([-30, 0, 40 , 50]) # m
field_data_Ghent_c = np.array([4.62*0.001, 8750*0.001, 45.21*0.001, 39.46*0.001]) # g/m^3
```

```python
plt.figure(figsize = [6,3])
plt.plot(field_data_Ghent_x, field_data_Ghent_c, color="black", marker = 'o', ms=10, ls = ':')
plt.xlabel('Location along Lieve Canal (relative to Oip-5) [m]')
plt.ylabel('Total concentration [g/m$^3$]')
plt.title('Observed contaminant concentrations at Ghent site')
plt.show()
```

## Modelling of baseline scenario
### Setup Data Input

```python
hydro = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.02, #[m]
    alpha_z = 0.004 #[m]
)
```

```python
att = mbt.AttenuationParameters(
    decay_rate = 0, #[1/day]
    retardation = 3.04 #[-]
)
```

```python
source = mbt.SourceParameters(
    source_zone_boundary = np.array([5,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)
```

```python
model = mbt.ModelParameters(
    model_length = 50, #[m]
    model_width = 120, #[m]
    model_time = 790, #[days]
    dx = 1, #[m]
    dy = 1, #[m]
    dt = 79, #[days]
)
```

### Modelling

```python
mbt_ghent = mbt.Mibitrans(hydro, att, source, model)
results_ghent = mbt_ghent.run()
```

### Plotting
#### Concentration along plume center over time

```python
cmap = plt.get_cmap("cubehelix")
colors = cmap(np.linspace(0.4, 0.7, 5))
```

```python
for i,plot_time in enumerate([7.9, 7, 5, 3, 1]):
    results_ghent.centerline(time=plot_time*100,
                               color=colors[i],
                               lw = 2,
                               ls = "-",
                               label=f"t={plot_time*100}days")
plt.title('Contaminant concentration along plume center')
plt.xlim([0,50])
plt.legend()
plt.show()
```

**Interpretation**: Over time, the contamination migrates further into the model domain, causing the concentration to increase. No steady state is reached, as the decay rate is set to $0$ $d^{-1}$.

#### Comparison to field measurements

```python
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: The field measurements align well with the mibitrans results. Although there are some minor differences, the results follow the same trend as the field measurements. However, due to the lack of field data between 0 and 30 meters on either side, and because of parameter uncertainties, there are multiple possible parameter combinations that can reproduce the field data.

## Scenario Modelling

### Dispersivity scenarios

```python
# Setup data input
hydro_d1 = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.1, #[m]
    alpha_z = 0.004 #[m]
)

hydro_d2 = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.001, #[m]
    alpha_z = 0.004 #[m]
)
```

**Note**: To run the scenarios, the entire model does not need to be recreated. It is sufficient to redefine the data class with the changed parameter(s).

```python
# Modelling
mbt_ghent_d1 = mbt.Mibitrans(hydro_d1, att, source, model)
results_ghent_d1 = mbt_ghent_d1.run()

mbt_ghent_d2 = mbt.Mibitrans(hydro_d2, att, source, model)
results_ghent_d2 = mbt_ghent_d2.run()
```

```python
# Plotting
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
results_ghent_d1.transverse(x_position= 40, lw = 2, ls = "-", color = colors[2],  label = "d1 scenario")
results_ghent_d2.transverse(x_position= 40, lw = 2, ls = "-", color = colors [4], label = "d2 scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: Although the shape of the concentration distribution differs across the three scenarios, they align with the field measurements in a similar manner. Thus, the "real" values of transverse horizontal dispersivity $\alpha_T$ can lie anywhere within this plausible range of 0.001 to 0.1 meters. Results show that the fit of the model to the field data is not sensitive to the chosen values for transverse horizontal dispersivity. 

### Source dimensions scenarios

```python
# Source data input
source_s0 = mbt.SourceParameters(
    source_zone_boundary = np.array([3,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)

source_s1 = mbt.SourceParameters(
    source_zone_boundary = np.array([15,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)

source_s2 = mbt.SourceParameters(
    source_zone_boundary = np.array([25,28]), #[m]
    source_zone_concentration = np.array([54.7,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)
```

```python
# Modelling
mbt_ghent_s0 = mbt.Mibitrans(hydro, att, source_s0, model)
results_ghent_s0 = mbt_ghent_s0.run()

mbt_ghent_s1 = mbt.Mibitrans(hydro, att, source_s1, model)
results_ghent_s1 = mbt_ghent_s1.run()

mbt_ghent_s2 = mbt.Mibitrans(hydro, att, source_s2, model)
results_ghent_s2 = mbt_ghent_s2.run()
```

```python
# Plotting
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
results_ghent_s0.transverse(x_position= 40, lw = 2, ls = "-", color = colors[1], label = "s0 scenario")
results_ghent_s1.transverse(x_position= 40, lw = 2, ls = "-", color = colors[3], label = "s1 scenario")
results_ghent_s2.transverse(x_position= 40, lw = 2, ls = "-", color = colors[4], label = "s2 scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: Although resulting concentration distributions among scenarios differ significantly, all provide a match to observed field data. Consequently, we cannot specify the width of the inner source zone, possibly ranging between 6 to 50 meters. 

### Combined variation scenario

```python
# Source data input
source_cv = mbt.SourceParameters(
    source_zone_boundary = np.array([4,10,28]), #[m]
    source_zone_concentration = np.array([62,20,2]), #[g/m3]
    total_mass = np.inf, #[g]
    depth = 15, #[m]
)

hydro_cv = mbt.HydrologicalParameters(
    velocity = 10.5/100, #[m/d]
    porosity = 0.3, #[-]
    alpha_x = 2.3, #[m]
    alpha_y = 0.2, #[m]
    alpha_z = 0.004 #[m]
)
```

```python
# Modelling
mbt_ghent_cv = mbt.Mibitrans(hydro_cv, att, source_cv, model)
results_ghent_cv = mbt_ghent_cv.run()
```

```python
#Plotting
results_ghent_cv.transverse(x_position= 40, lw = 2, ls = "-", color = colors[4], label = "cv scenario")
results_ghent.transverse(x_position= 40, lw = 2, ls = "-.", color = colors[0], label = "baseline scenario")
plt.scatter(field_data_Ghent_x, field_data_Ghent_c, color="black", label = "field measurements", s =50, zorder = 10)
plt.ylim(0,10)
plt.legend()
plt.show()
```

**Interpretation**: Again, scenario results differ from the concentration distribution calculated for the baseline scenario reflecting a stronger spreading perpendicular to the plume. Nevertheless, modelled concentrations still aligns with the field measurements. Various combinations of the uncertain parameters can also yield representative scenarios.



### Effect on concentration along plume center line

So far, we compared scenario results for the concentration distribution transverse to the flow direction due to the given observations at $x=40$m distance from the source. Find below the concentation distributions along the plume center line $x=0$ for selected scenarios compared to the baseline scenario.


```python
# Plotting
results_ghent.centerline(lw = 3, ls = "-", color = colors[0], label = "baseline scenario")
results_ghent_d1.centerline(lw = 2, ls = "--", color = colors[2],  label = "d1 scenario")
results_ghent_s0.centerline(lw = 2, ls = ":", color = colors[1], label = "s0 scenario")
results_ghent_cv.centerline(lw = 2, ls = "-", color = colors[4], label = "cv scenario")

plt.xlim(0,50)
plt.ylim(0,65)
plt.legend()
plt.show()
```

**Interpretation**: 
We see that only the combined variation scenario differs significantly in it's concentration distribution. The higher values of vertical spreading requires a higher source concentration to result in the same concentration values at $40$m distance to the source. However, differences fade away at larger distance from the source. Additional observations along the transport path, especially within the first 30 meters, would allow a better specification of the field situation.


## Summary and Conclusion

A simple *mibitrans* model with reasonably estimated flow and transport parameters was able to reproduce observed contaminant concentrations at the Ghent site. However, limited information create uncertainty regarding how representative the modelling results are for the field situation. We tested various parameter scenarios for the parameter values with the highest uncertainties, namely transverse dispersivity and source dimension/concentrations.

All tested scenarios could provide concentration distributions that are consistent the sparse field observations. A further specification of the model would require additional field data. However, when facing such uncertainties, *mibitrans* is well suited to rapidly determine the potential range of uncertain parameters and running scenarios - as demonstrated in this notebook.
